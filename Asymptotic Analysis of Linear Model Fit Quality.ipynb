## Asymptotic Analysis of LASSO Model Fit Quality

The LASSO regression model fits effective resistance $R_{\text{eff}}(u, v)$ as a function of geometric features. The key question we want to investigate is: **how does the quality of the model fit behave in the thermodynamic limit as $n \to \infty$?**

### Setup

We study fit quality (measured by out-of-sample $R^2$) across a grid of $(n, k)$ values, where:

- $n \in [256, 512, 1024, 2048, 4096]$ is the number of nodes
- $k \in [5, 6, 7, 8, 9]$ is the **expected mean degree**, which controls the connection radius via $r = \sqrt{k / ((n-1)\pi)}$

The model is trained by pooling samples from independently generated RGGs and evaluated on held-out graphs, ensuring the $R^2$ score reflects out-of-sample generalisation rather than in-sample memorisation.

### The Fixed Mean Degree Regime

In the **supercritical regime** used here, $k$ is held fixed as $n$ grows. Since the radius $r \propto n^{-1/2}$ shrinks as $n$ increases, each node's neighbourhood becomes a smaller and smaller fraction of the torus. Locally, the graph looks increasingly like an infinite planar random geometric graph.

This raises a natural question: does our fixed geometric feature set remain a good predictor of $R_{\text{eff}}$ in this super critical regime limit, or does the fit deteriorate as the graph grows large and the effective resistance landscape becomes harder to capture with low-order terms?

We expect the answer to depend on $k$:

The reference curve $\ln(n) = k$ marks the boundary of the **connected regime**: values of n above this curve make the supercritical radius no longer sufficient to guarantee connectivity almost surely. Points beyond this line should be interpreted with caution, as disconnected graphs are resolved by restricting to the largest connected component.

```python
import numpy as np
import networkx as nx
import random
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from RGG_Library import RGGBuilder
import pandas as pd

SPACE = "torus"
USE_ANGLES = True
PERTURB = False

# ---------------------------
# GRID PARAMETERS
# ---------------------------

N_VALUES            = [512, 1024, 2048, 4096, 8192]
K_VALUES            = [5, 6, 8, 10, 12, 16, 20]
NSAMPLES            = 1000
N_TEST_GRAPHS       = 25
N_TRAIN_GRAPHS      = 25
N_SHUFFLES          = 50
BASE_SEED           = 0
N_GRAPHS            = N_TRAIN_GRAPHS + N_TEST_GRAPHS

# ---------------------------
# HELPER: build graph & extract features
# ---------------------------

def build_features(n, k, nsamples, seed):
    builder = RGGBuilder(
        n=n, k=k, connectivity_regime="sc",
        space=SPACE, order=4, perturb=PERTURB, seed=seed
    )
    G = builder.build()

    if not nx.is_connected(G):
        components = list(nx.connected_components(G))
        if not components:
            return None, None
        Gsub = G.subgraph(max(components, key=len)).copy()
    else:
        Gsub = G

    Gsub = nx.convert_node_labels_to_integers(Gsub, ordering="sorted")

    random.seed(seed)
    np.random.seed(seed)

    if USE_ANGLES:
        res, degs, dists, pairs, angles = RGGBuilder.sample_commute_times_even_distance_w_angles(
            Gsub, nsamples=nsamples, n_bins=20, seed=seed,
            min_dist=builder.radius, max_dist=2
        )
    else:
        res, degs, dists, pairs = RGGBuilder.sample_commute_times_even_distance(
            Gsub, nsamples=nsamples, n_bins=20, seed=seed,
            min_dist=builder.radius, max_dist=2
        )

    if len(res) < 10:
        print(f"  [DROPPED] n={n}, k={k}, seed={seed}: only {len(res)} samples returned")
        return None, None

    safe_dists = dists.copy()
    safe_dists[safe_dists == 0] = 1e-9
    current_angles = angles if USE_ANGLES else np.zeros_like(dists)

    X = np.column_stack([
        dists**2,
        np.log(safe_dists),
        dists**4 * np.cos(4 * current_angles),
        degs,
    ])
    y = np.array(res)

    return X, y


# ---------------------------
# CORE FUNCTION
# ---------------------------

def compute_r2(n, k, nsamples=NSAMPLES, base_seed=BASE_SEED):
    try:
        graphs_X, graphs_y = [], []
        for i in range(N_GRAPHS):
            X, y = build_features(n, k, nsamples, seed=base_seed + i)
            if X is not None:
                graphs_X.append(X)
                graphs_y.append(y)

        n_graphs = len(graphs_X)
        print(f"  n={n}, k={k}: successfully built {n_graphs}/{N_GRAPHS} graphs")

        if len(graphs_X) < 2:
            return np.nan, np.nan

        r2_list   = []
        coef_list = []

        for i in range(N_SHUFFLES):
            rng = np.random.default_rng(seed=base_seed * 1000 + i)
            idx = rng.permutation(n_graphs)
            train_idx = idx[:N_TRAIN_GRAPHS]
            test_idx  = idx[N_TRAIN_GRAPHS:]

            if len(test_idx) == 0:
                r2_list.append(np.nan)
                coef_list.append([np.nan] * 5)
                continue

            X_train = np.vstack([graphs_X[j] for j in train_idx])
            y_train = np.concatenate([graphs_y[j] for j in train_idx])
            X_test  = np.vstack([graphs_X[j] for j in test_idx])
            y_test  = np.concatenate([graphs_y[j] for j in test_idx])

            model = LinearRegression()
            model.fit(X_train, y_train)

            real_coefs     = model.coef_
            real_intercept = model.intercept_

            y_pred = X_test @ real_coefs + real_intercept
            ss_res = np.sum((y_test - y_pred) ** 2)
            ss_tot = np.sum((y_test - np.mean(y_test)) ** 2)
            r2_list.append(1 - ss_res / ss_tot)
            coef_list.append([*real_coefs, real_intercept])

        return r2_list, coef_list

    except Exception as e:
        print(f"  [WARN] n={n}, k={k} failed: {e}")
        return np.nan, np.nan


# ---------------------------
# SWEEP OVER GRID
# ---------------------------

COEF_NAMES = ["coef_dist^2", "coef_log(dist)", "coef_4cos(4theta)", "coef_invdeg", "intercept"]

R2_grid   = np.full((len(N_VALUES), len(K_VALUES), N_SHUFFLES), np.nan)
Coef_grid = np.full((len(N_VALUES), len(K_VALUES), N_SHUFFLES, 5), np.nan)

for i, n in enumerate(N_VALUES):
    for j, k in enumerate(K_VALUES):
        print(f'N_VAL, K_VAL: {i}, {j}')
        r2_list, coef_list = compute_r2(n, k, nsamples=NSAMPLES, base_seed=BASE_SEED)
        if not np.isnan(r2_list).all():
            R2_grid[i, j, :]      = r2_list
            Coef_grid[i, j, :, :] = coef_list

# ---------------------------
# BUILD DATAFRAME
# ---------------------------

records = []
for i, n in enumerate(N_VALUES):
    for j, k in enumerate(K_VALUES):
        for s in range(N_SHUFFLES):
            record = {
                "n":               n,
                "k":               k,
                "ln(n)":           np.log(n),
                "shuffle":         s,
                "r2":              R2_grid[i, j, s],
                "above_threshold": k >= np.log(n),
            }
            for c, name in enumerate(COEF_NAMES):
                record[name] = Coef_grid[i, j, s, c]
            records.append(record)

df = pd.DataFrame(records)
df["r2"] = df["r2"].astype(float)

print("\nResults DataFrame:")
print(df.to_string(index=False))

# ---------------------------------------------------------------------------
# SAVE
# ---------------------------------------------------------------------------

CSV_PATH = "DataFrames/Assymptotic_Analysis_R2_Results_OLS.csv"
df.to_csv(CSV_PATH, index=False)
print(f"\nDataFrame saved to {CSV_PATH}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.lines as mlines

# ---------------------------------------------------------------------------
# FONT
# ---------------------------------------------------------------------------

plt.rcParams.update({
    "text.usetex":        False,
    "font.family":        "serif",
    "font.serif":         ["cmr10", "CMU Serif", "DejaVu Serif"],
    "mathtext.fontset":   "cm",
    "axes.formatter.use_mathtext": True,
    "axes.unicode_minus": False,
})
FS = 2.0
plt.rcParams["font.size"] = 10 * FS   # drives the main-axis tick labels

# ---------------------------------------------------------------------------
# LOAD DATA
# ---------------------------------------------------------------------------

df = pd.read_csv("DataFrames/Assymptotic_Analysis_R2_Results_OLS.csv")

# ---------------------------------------------------------------------------
# COLUMNS
# ---------------------------------------------------------------------------

metrics = {
    "Intercept":                                                 "intercept",
    r"$\frac{1}{\deg(u)}+\frac{1}{\deg(v)}$: Inverse Degree Sum": "coef_invdeg",
    r"$\ln(d)$: Logarithm of Distance":                        "coef_log(dist)",
    r"$d^2$: Background Correction":                            "coef_dist^2",
    r"$d^4\cos(4\theta)$: Angular Correction":                  "coef_4cos(4theta)",
    r"$R^2$":                                                    "r2",
}

# Columns that get a log-log inset:
#   key → (take_abs, [left, bottom, width, height] in axes fraction, shift)
INSET_COLS = {
    "intercept":          (False, [0.42, 0.35, 0.56, 0.56],  0),  # upper-right
    "coef_invdeg":        (False, [0.42, 0.35, 0.56, 0.56],  1),  # upper-right, plot coef−1
    "coef_log(dist)":     (False, [0.42, 0.35, 0.56, 0.56],  0),  # upper-right
    "coef_dist^2":        (True,  [0.42, 0.08, 0.56, 0.56],  0),  # bottom-right
    "coef_4cos(4theta)":  (True,  [0.42, 0.08, 0.56, 0.56],  0),  # bottom-right
}

# ---------------------------------------------------------------------------
# AGGREGATE
# ---------------------------------------------------------------------------

stats = (
    df.groupby(["n", "k"])[list(metrics.values())]
    .agg(["mean", "std"])
    .reset_index()
)

n_values = sorted(df["n"].unique())
k_values = sorted(df["k"].unique())
colors   = cm.viridis(np.linspace(0.1, 0.9, len(n_values)))

ln_n = {n: np.log(n) for n in n_values}

# 1/k^2 reference: k = 12 to 20 only
k_ref = np.linspace(12, 20, 200)
ref_y = 1.0 / k_ref**2

# ---------------------------------------------------------------------------
# ADD INSET
# ---------------------------------------------------------------------------

def add_inset(ax, col, take_abs, pos, shift=0):
    axins = ax.inset_axes(pos)

    for n, color in zip(n_values, colors):
        sub  = stats[stats["n"] == n].sort_values("k")
        mu   = sub[(col, "mean")]
        sd   = sub[(col, "std")]
        y    = (mu.abs() if take_abs else mu) - shift
        mask = y > 0
        if mask.sum() < 2:
            continue
        axins.errorbar(
            sub["k"][mask], y[mask], yerr=sd[mask],
            color=color, marker="o", markersize=3,
            linewidth=1.2, capsize=2, capthick=1.0, elinewidth=0.8,
        )

    all_y = []
    for n in n_values:
        sub = stats[stats["n"] == n].sort_values("k")
        mu  = sub[(col, "mean")]
        y   = (mu.abs() if take_abs else mu) - shift
        all_y.extend(y[y > 0].tolist())

    if all_y:
        y_center = np.quantile(all_y, 0.65)
        scale    = y_center / (1.0 / np.median(k_values) ** 2)
        axins.plot(k_ref, scale * ref_y, "k--", linewidth=1.2,
                   label=r"$\propto 1/k^2$")
        axins.legend(fontsize=6 * FS, framealpha=0.75, loc="upper right",
                     handlelength=1.5)

    axins.set_xscale("log")
    axins.set_yscale("log")
    axins.set_xticks(k_values)
    axins.get_xaxis().set_major_formatter(plt.ScalarFormatter())

    if shift:
        ylabel = r"$c_{1/d} - 1$" + " (log)"
    else:
        ylabel = (r"$|\mathrm{coef}|$" if take_abs else "value") + " (log)"
    axins.set_ylabel(ylabel, fontsize=9 * FS, labelpad=1)

    axins.tick_params(axis="both", labelsize=6 * FS)
    axins.grid(True, which="both", linestyle="--", alpha=0.35)
    axins.set_title("log-log", fontsize=9 * FS, pad=2, color="0.35")
    for spine in axins.spines.values():
        spine.set_edgecolor("0.45")
        spine.set_linewidth(0.8)

# ---------------------------------------------------------------------------
# MAIN FIGURE
# ---------------------------------------------------------------------------

ncols = 2
nrows = int(np.ceil(len(metrics) / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4.5 * nrows))
axes = axes.flatten()

legend_handles = []
legend_labels  = []

for ax, (label, col) in zip(axes, metrics.items()):
    for n, color in zip(n_values, colors):
        sub = stats[stats["n"] == n].sort_values("k")

        container = ax.errorbar(
            sub["k"],
            sub[(col, "mean")],
            yerr=sub[(col, "std")],
            color=color,
            marker="o",
            markersize=5,
            linewidth=1.6,
            capsize=4,
            capthick=1.3,
            elinewidth=1.1,
        )

        if len(legend_handles) < len(n_values):
            legend_handles.append(container.lines[0])
            legend_labels.append(f"n = {n}")

        ax.axvline(ln_n[n], color=color, linewidth=1.2, linestyle="--", alpha=0.7)

    ax.set_xlabel("k", fontsize=10 * FS)
    ax.set_ylabel(f"(mean $\\pm$ 1 std)", fontsize=10 * FS)
    ax.set_title(label, fontsize=12 * FS, fontweight="bold")
    ax.set_xticks(k_values)
    ax.grid(True, linestyle="--", alpha=0.4)

    if col in INSET_COLS:
        take_abs, pos, shift = INSET_COLS[col]
        add_inset(ax, col, take_abs, pos, shift)

for ax in axes[len(metrics):]:
    ax.set_visible(False)

plt.tight_layout()

# ---------------------------------------------------------------------------
# LEGEND
# ---------------------------------------------------------------------------

vline_handle = mlines.Line2D(
    [], [], color="gray", linewidth=1.2, linestyle="--", label="k = ln(n)"
)
legend_handles.append(vline_handle)
legend_labels.append("k = ln(n)")

leg = fig.legend(
    legend_handles, legend_labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0),
    ncol=len(legend_handles),
    fontsize=10 * FS,
    framealpha=0.85,
    edgecolor="0.7",
)

plt.savefig("Figures/all_coeffs_vs_k_insets.pdf", bbox_inches="tight")
print("Saved.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------------------------
# FONT
# ---------------------------------------------------------------------------

plt.rcParams.update({
    "text.usetex":        False,
    "font.family":        "serif",
    "font.serif":         ["cmr10", "CMU Serif", "DejaVu Serif"],
    "mathtext.fontset":   "cm",
    "axes.formatter.use_mathtext": True,
    "axes.unicode_minus": False,
})
FS = 2.0
plt.rcParams["font.size"] = 7 * FS   # drives the main-axis tick labels

# ---------------------------------------------------------------------------
# LOAD DATA
# ---------------------------------------------------------------------------

df = pd.read_csv("DataFrames/Assymptotic_Analysis_R2_Results_OLS.csv")

# ---------------------------------------------------------------------------
# COLUMNS
# ---------------------------------------------------------------------------

metrics = {
    "Intercept":                                                 "intercept",
    r"$\frac{1}{\deg(u)}+\frac{1}{\deg(v)}$: Inverse Degree Sum": "coef_invdeg",
    r"$\ln(d)$: Logarithm of Distance":                        "coef_log(dist)",
    r"$d^2$: Background Correction":                            "coef_dist^2",
    r"$d^4\cos(4\theta)$: Angular Correction":                  "coef_4cos(4theta)",
    r"$R^2$":                                                    "r2",
}

# ---------------------------------------------------------------------------
# AGGREGATE k=20
# ---------------------------------------------------------------------------

stats = (
    df[df["k"] == 20]
    .groupby("n")[list(metrics.values())]
    .agg(["mean", "std"])
    .reset_index()
)
stats["inv_n"] = 1 / stats["n"]
stats = stats.sort_values("inv_n")

n_values       = sorted(df["n"].unique())
tick_positions = [1 / n for n in n_values]
tick_labels    = [f"n={n}" for n in n_values]

# ---------------------------------------------------------------------------
# PLOT
# ---------------------------------------------------------------------------

ncols = 2
nrows = int(np.ceil(len(metrics) / ncols))

fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4.5 * nrows))
axes = axes.flatten()

for ax, (label, col) in zip(axes, metrics.items()):
    mean = stats[(col, "mean")].to_numpy()
    std  = stats[(col, "std")].to_numpy()
    x    = stats["inv_n"].to_numpy()

    ax.plot(
        x,
        mean,
        color="steelblue",
        marker="o",
        markersize=6,
        linewidth=1.8,
    )
    ax.fill_between(
        x,
        mean - std,
        mean + std,
        color="steelblue",
        alpha=0.15,
    )

    for xv in tick_positions:
        ax.axvline(xv, color="grey", linewidth=0.5, linestyle="--", alpha=0.4)

    ax.set_xticks(tick_positions)
    ax.set_xticklabels(tick_labels, rotation=40, ha="right", fontsize=8.5 * FS)
    ax.set_xlim(left=0)
    ax.set_ylabel(f"(mean $\\pm$ 1 std)", fontsize=10 * FS)
    ax.set_title(label, fontsize=12 * FS, fontweight="bold")
    ax.grid(True, linestyle="--", alpha=0.3)

    # Log-log inset for Intercept only
    if col == "intercept":
        ax_ins = ax.inset_axes([0.53, 0.50, 0.44, 0.46])

        ax_ins.plot(
            x,
            mean,
            color="steelblue",
            marker="o",
            markersize=4,
            linewidth=1.4,
        )
        ax_ins.fill_between(
            x,
            mean - std,
            mean + std,
            color="steelblue",
            alpha=0.15,
        )
        ax_ins.set_xscale("log")
        ax_ins.set_xticks(tick_positions)
        ax_ins.get_xaxis().set_major_formatter(
            plt.FuncFormatter(lambda val, _: f"n={int(round(1/val))}" if val > 0 else "")
        )
        ax_ins.tick_params(axis="x", labelrotation=40, labelsize=5.5 * FS)
        ax_ins.tick_params(axis="y", labelsize=5.5 * FS)
        ax_ins.set_ylabel("Intercept", fontsize=6.5 * FS)
        ax_ins.grid(True, which="both", linestyle="--", alpha=0.3)
        ax_ins.patch.set_facecolor("white")
        ax_ins.patch.set_alpha(0.9)
        for spine in ax_ins.spines.values():
            spine.set_edgecolor("0.45")
            spine.set_linewidth(0.8)

# ---------------------------------------------------------------------------
# PLOT
# ---------------------------------------------------------------------------

for ax in axes[len(metrics):]:
    ax.set_visible(False)

plt.tight_layout()

plt.savefig("Figures/coeffs_vs_inv_n_k20.pdf", bbox_inches="tight")
print("Saved.")
```
